```python
def fill_null_valuese(data_frame_name):
    
    for column in data_frame_name.columns:
        
        if data_frame_name[column].isnull().sum() > 0:
            
            if data_frame_name[column].dtypes in ['int64', 'int32', 'float64', 'float32' ]:
                data_frame_name[column].fillna(value = data_frame_name[column].mean(), inplace = True )
                
            else :
                data_frame_name[column].fillna(value = data_frame_name[column].mode()[0], inplace = True )
                
    return print( f"Fill Null values  operaction has been Done \n{data_frame_name.isnull().sum()}")    
```

```python
fill_null_valuese(data)
```
